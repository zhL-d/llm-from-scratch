```python
from cs336_basics.tokenizer import Tokenizer
import time
```

```python
def pretty_output(corpus: str, token_ids: list[int], preview_len: int, elapsed_time: float | None =None):
    corpus_bytes = corpus.encode()

    print("[Tokenizer] encode complete")
    print(f"    -Corpus length (bytes): {len(corpus_bytes)}")
    print(f"    -Token count: {len(token_ids)}")
    print(f"    -Compression ratio: {len(corpus_bytes) / len(token_ids)}")

    if elapsed_time:
        print(f"    -Elapsed time: {elapsed_time:.4f} seconds")
        print(f"    -Throughput: {len(corpus_bytes) / elapsed_time} bytes/second")

    print(f"    - Preview (corpus): {corpus[:preview_len]}...")

    print(f"    - Preview (token ids): {token_ids}")
```

```python
vocab_path = "/Users/lucas/Documents/GitHub/stf-assignment1-basics/cs336_basics/smoke_test_fixture/vocab_table/tiny_story/output_TinyStoriesV2-GPT4-train_serialization_vocab_20251010_112414.json"
merge_path = "/Users/lucas/Documents/GitHub/stf-assignment1-basics/cs336_basics/smoke_test_fixture/vocab_table/tiny_story/output_TinyStoriesV2-GPT4-train_serialization_merge_20251010_112414.json"

corpus_path = "/Users/lucas/Documents/GitHub/stf-assignment1-basics/cs336_basics/smoke_test_fixture/sample_data/sample_TinyStoriesV2-GPT4-train_k10.txt"

# ---- Start timing ----

start_time = time.perf_counter()

tokenizer = Tokenizer.from_files(vocab_path, merge_path, ["<|endoftext|>"])

with open(corpus_path) as f_corpus:
    corpus = f_corpus.read()

token_ids = tokenizer.encode(corpus)

# ---- End timing ----
end_time = time.perf_counter()
elapsed_time = end_time - start_time

# print(f"[Tokenizer] encodeing complete in {elapsed_time:.4f} seconds")
pretty_output(corpus, token_ids, 30, elapsed_time)
```

**Answer:** the TinyStories tokenizer’s compression ratio is 4.312151616499443

**Answer:** Throughput: 310.6875414523819 bytes/second;  It would take 2819386382.2198973 seconds to tokenize the Pile dataset (825GB of text), equivalent to 32631.78683124881 days, equivalent to 89.40215570205153 years
