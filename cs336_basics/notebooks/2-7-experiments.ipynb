```python
def pretty_output(corpus: str, token_ids: list[int], preview_len: int):
    corpus_bytes = corpus.encode()

    print("[Tokenizer] encode complete")
    print(f"    -Corpus length (bytes): {len(corpus_bytes)}")
    print(f"    -Token count: {len(token_ids)}")
    print(f"    -Compression ratio: {len(corpus_bytes) / len(token_ids)}")

    print(f"    - Preview (corpus): {corpus[:preview_len]}...")

    print(f"    - Preview (token ids): {token_ids}")
```
